````python
class Calculator:
    class Arithmetic:
        def add(self, a: float, b: float) -> float:
            return a + b

        def subtract(self, a: float, b: float) -> float:
            return a - b

        def multiply(self, a: float, b: float) -> float:
            return a * b

        def divide(self, a: float, b: float) -> float:
            if b == 0:
                raise ValueError("Cannot divide by zero")
            return a / b

    class Advanced:
        def root(self, a: float, n: int) -> float:
            if n == 0:
                raise ValueError("Degree cannot be zero")
            if a < 0 and n % 2 == 0:
                raise ValueError("Even root of a negative number is not real")
            return math.copysign(abs(a) ** (1.0 / n), a)

        def power(self, a: float, b: float) -> float:
            return a ** b

        def sin_value(self, angle_deg: float) -> float:
            return math.sin(math.radians(angle_deg))

        def cos_value(self, angle_deg: float) -> float:
            return math.cos(math.radians(angle_deg))

        def tan_value(self, angle_deg: float) -> float:
            return math.tan(math.radians(angle_deg))

        def log_value(self, value: float, base: float = math.e) -> float:
            if value <= 0:
                raise ValueError("Log input must be positive")
            if base <= 0 or base == 1:
                raise ValueError("Base must be positive and not equal to 1")
            return math.log(value, base)


calc = Calculator()
art = calc.Arithmetic()
adv = calc.Advanced()

# Example checks
assert art.add(5, 3) == 8
assert art.subtract(10, 4) == 6
assert art.multiply(6, 7) == 42
assert art.divide(8, 2) == 4
assert adv.power(2, 3) == 8
assert round(adv.sin_value(30), 6) == 0.5
assert round(adv.log_value(100, 10), 6) == 2.0

print("Calculator functions are working correctly.")
````
